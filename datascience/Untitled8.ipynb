```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
df=pd.read_csv("data.csv")

x=df.iloc[:, :-1].values
y=df.iloc[:, -1].values
x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.2,random_state=42)
k=3
knn=KNeighborsClassifier(n_neighbors=k)
knn.fit(x_train,y_train)
y_pred=knn.predict[x_test]
Duration=float(input("enter the duration:"))
Pulse=float(input("enter the pulse:"))
Maxpulse=float(input("enter the maxpulse:"))
new_arr=([Duration,Pulse,Maxpulse])
new=knn.predict(new_arr)
accuracy=acuracy_score(y_test,new_arr)
print("accuracy:",accuracy,"predict:",new[0])
```
